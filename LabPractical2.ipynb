8. Creating a data frame

```r
students <- data.frame(
  Student_ID = c("S001", "S002", "S003", "S004", "S005"),
  Attendance = c(95, 72, 88, 79, 68),
  Grade = c(90, 70, 85, 77, 65)

)

students
```

9. Adding a Classification Column

```r
students$Performance <- ifelse(
  students$Grade >= 80,
  "Good",
  ifelse(
    students$Grade >= 75,
    "Passed",
    "Failed"
  )
)

students
```

10. Creating a Simple Rule-Based System

```r
students_risk <- function(attendance, grade) {
  if (attendance > 75 && grade > 75) {
    print("High Risk")
    } else {
      print("Low Risk")
      }
}

students
```

11. Apply the rule to multiple students

```r
students$Risk <- mapply(
  students_risk,
  students$Attendance,
  students$Grade
)

students
```

12. Count the Result

```r
table(students$Risk)
```

13. Create Visualization

```r
risk_counts <- table(students$Risk)

barplot(
  risk_counts,
  main = "Student Risk Classification",
  xlab = "Risk Levels",
  ylab = "Number of students"
)
```

14. Add Recommendation

```r
students$Recommendation <- ifelse(
  students$Risk == "High Risk",
  "Immediate academic intervention",
  ifelse(
    students$Risk == "Moderate Risk",
    "Academic monitoring",
    "Continue regular monitoring"
  )
)

students
```
