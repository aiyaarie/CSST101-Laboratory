# Table 1. Debate Arguments

```r
debate <- data.frame(
  ArgumentID = c("A001", "A002", "A003", "A004", "A005", "A006"),
  Position = c("For", "For", "For", "Against", "Against", "Against"),
  Topic = c("Learning", "Productivity", "AI Literacy", "Academic Integrity", "Dependency", "Assessment"),
  Claim = c("AI can provide students with immediete learning support.", "AI can help students complete routing academic tasks more efficiently.", "Allowing AI use can help students develop AI literacy.", "Undisclosed AI-generated work may create academic integrity concerns.", "Excessive AI use may reduce students's independent problem-solving practice.", "AI use may make it more difficult to determine whether submitted work reflects student learning.")

)

debate
```

# Table 2. Evidence

```r
evidence <- data.frame(
  EvidenceID = c("E001", "E002", "E003", "E004", "E005", "E006"),
  ArgumentID = c("A001", "A002", "A003", "A004", "A005", "A006"),
  EvidenceType = c("Research", "Research", "Policy", "Policy", "Research", "Research"),
  Strength = c("Strong", "Moderate", "Strong", "Strong", "Moderate", "Strong")

)

evidence
```

```r
table(debate$Position)
```

```r
nrow(debate)
```

# Connect Table of Debate and Evidence

```r
knowledge_base <- merge(
  debate,
  evidence,
  by = "ArgumentID"
)

print(knowledge_base)
```

# Convert Evidence Strength into a Score

```r
get_score <- function(Strength)(

  if(Strength == "Strong"){
    return(3)
  } else if (Strength == "Moderate"){
    return(2)
  } else if (Strength == "Weak"){
    return(1)
  } else {
    return(0)
  }
)

get_score("Moderate")
```

# Add Evidence Scores

```r
knowledge_base$EvidenceScore <- sapply(
  knowledge_base$Strength,
  get_score
)

print(knowledge_base)
```

# Convert Evidence Score to Support

```r
classify_support <- function(score){

   if(score == 3){
    return("Strongly Supported")
  } else if (score == 2){
    return("Moderately Supported")
  } else if (score == 1){
    return("Weakly Supported")
  } else {
    return("No Support")
  }
}

classify_support(3)
```

# Add Evidence Support

```r
knowledge_base$EvidenceSupport <- sapply(
  knowledge_base$EvidenceScore,
  classify_support
)

print(knowledge_base)
```

# Search by Topic

```r
search_topic <- function(topic) {

  result <- subset(
    knowledge_base,
    Topic == topic
  )

  if(nrow(result) == 0){
    cat("Topic not found")
  } else{
    print(result)
  }
}

search_topic("Learning")
```
